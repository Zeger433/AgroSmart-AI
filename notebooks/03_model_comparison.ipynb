# 03 — Model comparison
Reads `artifacts/models/*_metadata.json` after training and visualises leaderboard metrics.

```python
import json
from pathlib import Path
import pandas as pd
root = Path('..')
with open(root / 'artifacts/models/crop_metadata.json') as f:
    c = json.load(f)
with open(root / 'artifacts/models/yield_metadata.json') as f:
    y = json.load(f)
print('Best crop model:', c.get('best_model'))
print('Best yield model:', y.get('best_model'))
display(pd.DataFrame(c.get('models', {})).T)
display(pd.DataFrame(y.get('models', {})).T)
```
